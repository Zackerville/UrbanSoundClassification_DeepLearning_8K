```python
from pathlib import Path
import json
import random

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import librosa
import librosa.display
import soundfile as sf

from IPython.display import Audio, display
from tqdm.notebook import tqdm
```

```python
ROOT_DIR = Path.cwd()
if not (ROOT_DIR / "data").exists():
    ROOT_DIR = ROOT_DIR.parent

RAW_DATA_DIR = ROOT_DIR / "data" / "raw"
AUDIO_DIR = RAW_DATA_DIR / "audio"

METADATA_CANDIDATES = [
    ROOT_DIR / "data" / "metadata" / "UrbanSound8K.csv",
    RAW_DATA_DIR / "metadata" / "UrbanSound8K.csv",
]

METADATA_PATH = None
for path in METADATA_CANDIDATES:
    if path.exists():
        METADATA_PATH = path
        break

if METADATA_PATH is None:
    raise FileNotFoundError("Không tìm thấy UrbanSound8K.csv")

FIGURES_DIR = ROOT_DIR / "outputs" / "figures"
EDA_DIR = ROOT_DIR / "outputs" / "eda"

FIGURES_DIR.mkdir(parents=True, exist_ok=True)
EDA_DIR.mkdir(parents=True, exist_ok=True)

ROOT_DIR, AUDIO_DIR, METADATA_PATH
```

READ METADATA

```python
df = pd.read_csv(METADATA_PATH)
df.head()
```

```python
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
display(df.isna().sum().to_frame("missing_count"))
df.info()
```

CLASS AND FOLD DISTRIBUTION

```python
print("Number of classes:", df["class"].nunique())
print("Classes:", sorted(df["class"].unique().tolist()))

class_counts = df["class"].value_counts().sort_index()
display(class_counts.to_frame("count"))

plt.figure(figsize=(12, 6))
class_counts.plot(kind="bar")
plt.title("Number of Samples per Class")
plt.xlabel("Class")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(FIGURES_DIR / "class_distribution.png")
plt.show()
```

```python
fold_counts = df["fold"].value_counts().sort_index()
display(fold_counts.to_frame("count"))

plt.figure(figsize=(8, 5))
fold_counts.plot(kind="bar")
plt.title("Number of Samples per Fold")
plt.xlabel("Fold")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig(FIGURES_DIR / "fold_distribution.png")
plt.show()

pd.crosstab(df["class"], df["fold"])
```

CHECK MAPPING FROM METADATA TO AUDIO FILE

```python
df["audio_path"] = df.apply(
    lambda row: AUDIO_DIR / f"fold{int(row['fold'])}" / row["slice_file_name"],
    axis=1
)
df["file_exists"] = df["audio_path"].apply(lambda x: x.exists())

display(df["file_exists"].value_counts().to_frame("count"))
display(df.loc[~df["file_exists"], ["slice_file_name", "fold", "audio_path"]].head(20))
```

AUDIO INFORMATION: SAMPLE RATE, CHANNELS, DURATION

```python
audio_info_records = []

for _, row in tqdm(df.iterrows(), total=len(df)):
    if row["file_exists"]:
        info = sf.info(str(row["audio_path"]))
        audio_info_records.append({
            "slice_file_name": row["slice_file_name"],
            "fold": int(row["fold"]),
            "sample_rate": info.samplerate,
            "channels": info.channels,
            "duration_actual": info.duration,
            "frames": info.frames,
            "format": info.format,
            "subtype": info.subtype,
        })

audio_info_df = pd.DataFrame(audio_info_records)
audio_info_df.head()
```

```python
df = df.merge(audio_info_df, on=["slice_file_name", "fold"], how="left")

if "start" in df.columns and "end" in df.columns:
    df["duration_metadata"] = df["end"] - df["start"]
else:
    df["duration_metadata"] = np.nan

display(df[["sample_rate", "channels", "duration_actual", "duration_metadata"]].describe())

print("Unique sample rates:", sorted(df["sample_rate"].dropna().unique().tolist()))
print("Unique channels:", sorted(df["channels"].dropna().unique().tolist()))
```

```python
plt.figure(figsize=(10, 5))
df["sample_rate"].value_counts().sort_index().plot(kind="bar")
plt.title("Sample Rate Distribution")
plt.xlabel("Sample Rate")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig(FIGURES_DIR / "sample_rate_distribution.png")
plt.show()

plt.figure(figsize=(6, 4))
df["channels"].value_counts().sort_index().plot(kind="bar")
plt.title("Channel Distribution")
plt.xlabel("Channels")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig(FIGURES_DIR / "channel_distribution.png")
plt.show()

plt.figure(figsize=(8, 5))
df["duration_actual"].hist(bins=40)
plt.title("Duration Distribution")
plt.xlabel("Duration (seconds)")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig(FIGURES_DIR / "duration_distribution.png")
plt.show()
```

```python
if "salience" in df.columns:
    salience_map = {1: "foreground", 2: "background"}
    df["salience_name"] = df["salience"].map(salience_map).fillna(df["salience"].astype(str))
    display(df["salience_name"].value_counts().to_frame("count"))

    plt.figure(figsize=(6, 4))
    df["salience_name"].value_counts().plot(kind="bar")
    plt.title("Salience Distribution")
    plt.xlabel("Salience")
    plt.ylabel("Count")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.savefig(FIGURES_DIR / "salience_distribution.png")
    plt.show()
```

ILLUSTRATE SAMPLE AUDIO

```python
sample_row = df[df["file_exists"]].sample(1, random_state=42).iloc[0]
sample_row[["slice_file_name", "class", "fold", "sample_rate", "channels", "duration_actual", "audio_path"]]
```

```python
y, sr = librosa.load(sample_row["audio_path"], sr=None, mono=True)

plt.figure(figsize=(14, 4))
librosa.display.waveshow(y, sr=sr)
plt.title(f"Waveform - {sample_row['class']}")
plt.tight_layout()
plt.savefig(FIGURES_DIR / "sample_waveform.png")
plt.show()
```

```python
mel = librosa.feature.melspectrogram(y=y, sr=sr, n_fft=1024, hop_length=512, n_mels=128)
mel_db = librosa.power_to_db(mel, ref=np.max)

plt.figure(figsize=(14, 5))
librosa.display.specshow(mel_db, sr=sr, x_axis="time", y_axis="mel")
plt.colorbar(format="%+2.0f dB")
plt.title(f"Mel Spectrogram - {sample_row['class']}")
plt.tight_layout()
plt.savefig(FIGURES_DIR / "sample_mel_spectrogram.png")
plt.show()
```

SUMMARY SAVE

```python
summary = {
    "num_samples": int(len(df)),
    "num_classes": int(df["class"].nunique()),
    "classes": sorted(df["class"].unique().tolist()),
    "num_folds": int(df["fold"].nunique()),
    "folds": sorted(df["fold"].unique().tolist()),
    "all_files_exist": bool(df["file_exists"].all()),
    "sample_rates": {str(k): int(v) for k, v in df["sample_rate"].value_counts().sort_index().to_dict().items()},
    "channels": {str(k): int(v) for k, v in df["channels"].value_counts().sort_index().to_dict().items()},
    "duration_mean": float(df["duration_actual"].mean()),
    "duration_min": float(df["duration_actual"].min()),
    "duration_max": float(df["duration_actual"].max()),
}

with open(EDA_DIR / "summary.json", "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2, ensure_ascii=False)

df[[
    "slice_file_name",
    "fold",
    "class",
    "classID",
    "salience",
    "sample_rate",
    "channels",
    "duration_actual",
    "audio_path",
    "file_exists",
]].to_csv(EDA_DIR / "metadata_with_audio_info.csv", index=False)

summary
```
